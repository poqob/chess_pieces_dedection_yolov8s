The considered dataset must have a series of specific files for this project. Data such as <strong>Train Test Validation</strong> as well as the <strong>Yaml</strong> file, which I will give the necessary explanations about the importance of this file.
    
In the cell below you can see the <strong>Root_dir</strong> is the main root of dataset. Then we can join other <strong>materials</strong> with this Root for choosing our <strong>dependencies</strong>.

```python
#Root_dir = '/kaggle/input/chess-pieces-detection-image-dataset/Chess_pieces'
Root_dir = '/datasets/Chess_pieces'
```

<div style = 'border : 4px solid non; background-color:#f2f2f2 ; ;padding:10px'>
    
<strong>YAML is a human-readable data serialization format</strong> that is commonly used for configuration files. In the context of YOLOv8, YAML files are used to store the configuration settings for the model, such as the <strong>number of classes, the input image size, and the anchor boxes.</strong>

The <strong>yolov8.yaml</strong> file is the default configuration file for YOLOv8. It contains the default values for all the hyperparameters and settings used by the model. The file is divided into several sections, each of which corresponds to a different aspect of the model’s configuration.

For example, the model section contains information about the architecture of the model, such as the number of layers and the size of the filters. The train section contains information about the training process, such as the learning rate and the batch size. The data section contains information about the dataset, such as the location of the training and validation data.

In summary, <strong>YAML files are used to store the configuration settings for YOLOv8</strong>, and the yolov8.yaml file is the default configuration file for the model.

```python
yaml_path = os.path.join(Root_dir, '/data.yaml.yaml')
print(Root_dir+yaml_path)
```

```python
train_path = os.path.join(Root_dir, 'train', 'images')
valid_path = os.path.join(Root_dir, 'valid', 'images') #TODO: what is valid?
```

<div style = 'border : 4px solid non; background-color:#f2f2f2 ; ;padding:10px'>
    
In the cell below, the <strong>model.train</strong> is the key training function that can <strong>tune by changing the methods like epoch, batch, learning rate</strong> and etc.

* **[Table of Content](#0)**

```python
results = model.train(
    data="D:\\Dosyalar\\projeler\\py\\chess_yolo\\"+Root_dir+yaml_path,
    epochs=15,
    batch=32,
    lr0=0.0001, 
    lrf=0.1,
    imgsz=640,
    plots= True
)
```

<div style = 'border : 4px solid non; background-color:#f2f2f2 ; ;padding:10px'>

<strong>Lets see some of our outputs:<strong>
    

```python
data_dir = "D:\\Dosyalar\\projeler\\py\\chess_yolo\\"+"runs\\detect\\train10"
img_dir = os.path.join(data_dir, '*g')

files = glob.glob(img_dir)
#print(files.__len__())

imgs = []
for image in files:
    img = cv2.imread(image)
    imgs.append(img)
    plt.figure(figsize=(10,10), dpi= 200)
    plt.imshow(img)
```
